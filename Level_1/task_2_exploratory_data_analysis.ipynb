```python
#IMPORTING LIBRARIES
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
#Loading the cleaned iris dataset from task 1
df = pd.read_csv('iris_cleaned.csv')
df
```

```python
#Exploring the dataset
df.dtypes

```

```python
#Univariate Analysis
df['species'].value_counts()
```

```python
#Sorting the dataset based on sepal_length in descending order
df.sort_values(by= 'sepal_length', ascending= False)
```

```python
#selecting rows where petal_length is greater than 1.5
df[df['petal_length'] >1.5]
```

```python
#checking for duplicate rows in the dataset
df[df.duplicated()==True]
```

```python
#finding the mean of each species
df.groupby('species').mean()
```

```python
#   Plotting an histogram of the sepal length column
sns.histplot(data = df, x='sepal_length', color = 'red', kde=True)
plt.title('Distribution Of Sepal Length')
plt.show()
```

```python
#Plotting a boxplot of the sepal length column
sns.boxplot(data = df, x='sepal_length')
plt.title('Boxplot of Sepal Length')
plt.show()
```

```python
#Plotting a pairplot of the dataset
sns.pairplot(df, hue='species')
plt.title('Pairplot of the Dataset')
plt.show()
```

```python
#Plotting a distribution plot of the sepal length column for each species
sns.displot( data = df, x='sepal_length', col='species', kde=True, color = 'green')
plt.title('Distribution of Sepal Length for Each Species')
plt.show()
```

```python
#Plotting a distribution plot of the sepal width column for each species
sns.displot(data = df, x='sepal_width', kind='kde', col='species', color = 'green')
plt.title('Distribution of Sepal Width for Each Species')
plt.show()
```

```python
#Cross-checking the column names of the dataset
df.columns
```

```python
#Scatter plot of sepal length and sepal width
sns.scatterplot(data = df, x='sepal_length', y='sepal_width', hue='species')
plt.title('Sepal Length vs Sepal Width')
plt.show()
```

```python
# Calculate correlation matrix for numeric columns
corr_matrix = df.drop(columns=['species']).corr()
print(corr_matrix)

# Visualize it using a Seaborn heatmap
sns.heatmap(corr_matrix, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```
